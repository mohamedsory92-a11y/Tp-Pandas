```python
import pandas as pd
import numpy as np

# Загрузка заказов
order_files = [
    "data/orders_2024_01.csv",
    "data/orders_2024_02.csv",
    "data/orders_2024_03.csv"
]

df_orders = pd.concat(
    [pd.read_csv(f) for f in order_files],
    ignore_index=True
)

# Загрузка товаров и категорий
sheets = pd.read_excel(
    "data/products.xlsx",
    sheet_name=["products", "categories"]
)

df_products = sheets["products"]
df_categories = sheets["categories"]

# Загрузка событий
df_events_records = pd.read_json(
    "data/events_records.json",
    orient="records"
)

df_events_lines = pd.read_json(
    "data/events_lines.json",
    lines=True
)

df_events = pd.concat(
    [df_events_records, df_events_lines],
    ignore_index=True
)

df_events["event_time"] = pd.to_datetime(df_events["event_time"])

# Объединение товаров и категорий
df_prod_full = df_products.merge(
    df_categories,
    on="category_id",
    how="left",
    suffixes=("", "_cat")
)

print("Объединённая таблица товаров:")
print(df_prod_full.head())

# Объединение заказов с товарами и категориями
df_orders_prod = df_orders.merge(
    df_prod_full,
    on="product_id",
    how="left"
)

print("Заказы с информацией о товарах:")
print(df_orders_prod.head())

# Анализ заказов по категориям
category_stats = (
    df_orders_prod
    .groupby("category_name")
    .agg(
        orders_count=("order_id", "nunique"),
        users_count=("user_id", "nunique"),
        total_quantity=("quantity", "sum"),
        total_revenue=("price", lambda x: x.sum())
    )
    .reset_index()
)

print("Статистика по категориям:")
print(category_stats)

# Объединение событий с данными о товарах по product_id,
# если идентификатор товара присутствует в событиях
if "product_id" in df_events.columns:
    df_events_products = df_events.merge(
        df_prod_full,
        on="product_id",
        how="left"
    )
    print("События с информацией о товарах:")
    print(df_events_products.head())
else:
    print("В событиях нет столбца product_id.")

# Сохранение результатов
df_prod_full.to_csv(
    "TP6_Variant9_products.csv",
    index=False,
    encoding="utf-8-sig"
)

df_orders_prod.to_csv(
    "TP6_Variant9_orders_products.csv",
    index=False,
    encoding="utf-8-sig"
)

category_stats.to_csv(
    "TP6_Variant9_category_stats.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Результаты TP6 сохранены.")
```
