```python
np.random.seed(42)

n = 2000
statuses = ["delivered", "cancelled", "pending", "returned"]

df_orders = pd.DataFrame({
    "order_id": range(1, n+1),
    "order_datetime": pd.to_datetime("2024-01-01") + pd.to_timedelta(np.random.randint(0, 90*24*60, n), unit="m"),
    "price": np.round(np.random.exponential(scale=1500, size=n) + 200, 2),
    "quantity": np.random.randint(1, 6, n),
    "status": np.random.choice(statuses, n, p=[0.6, 0.15, 0.15, 0.1])
})
df_orders["revenue"] = df_orders["price"] * df_orders["quantity"]
print(df_orders.head())
print(df_orders.shape)
```

```python
df_orders["order_date"] = df_orders["order_datetime"].dt.date
df_orders["order_hour"] = df_orders["order_datetime"].dt.hour
df_orders["weekday"] = df_orders["order_datetime"].dt.day_name()
print(df_orders.head())
```

```python
df_orders["price"].hist(bins=30)
plt.title("Распределение price")
plt.xlabel("Price")
plt.ylabel("Количество заказов")
plt.show()
```

```python
sns.boxplot(x=df_orders["price"])
plt.title("Boxplot price")
plt.xlabel("Price")
plt.show()
```
