```python
import pandas as pd
import numpy as np

df_events_records = pd.read_json("data/events_records.json", orient="records")
df_events_lines = pd.read_json("data/events_lines.json", lines=True)

df_events = pd.concat([df_events_records, df_events_lines], ignore_index=True)

df_events["event_time"] = pd.to_datetime(df_events["event_time"])

df_events["weekday"] = df_events["event_time"].dt.day_name()

days = {
    "Monday": "Понедельник",
    "Tuesday": "Вторник",
    "Wednesday": "Среда",
    "Thursday": "Четверг",
    "Friday": "Пятница",
    "Saturday": "Суббота",
    "Sunday": "Воскресенье"
}

df_events["weekday"] = df_events["weekday"].map(days)

print(df_events.head())

pivot_table = pd.crosstab(df_events["weekday"], df_events["event_type"])

print("Сводная таблица событий:")
print(pivot_table)

event_shares = pivot_table.div(pivot_table.sum(axis=1), axis=0)

print("Доля каждого типа событий по дням недели:")
print(event_shares.round(4))

if "add_to_cart" in event_shares.columns:
    max_day = event_shares["add_to_cart"].idxmax()
    max_share = event_shares["add_to_cart"].max()
    print("День недели с максимальной долей add_to_cart:", max_day)
    print("Максимальная доля:", round(max_share * 100, 2), "%")
else:
    print("События add_to_cart в данных отсутствуют.")

pivot_table.to_csv("TP4_Variant9_pivot.csv", encoding="utf-8-sig")
event_shares.to_csv("TP4_Variant9_shares.csv", encoding="utf-8-sig")

print("Результаты успешно сохранены.")
```
