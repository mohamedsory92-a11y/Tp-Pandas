```python
import pandas as pd
import matplotlib.pyplot as plt

# Загрузка данных

orders_files = [
    "data/orders_2024_01.csv",
    "data/orders_2024_02.csv",
    "data/orders_2024_03.csv"
]

df_orders = pd.concat(
    [pd.read_csv(f) for f in orders_files],
    ignore_index=True
)

# 1. Столбчатая диаграмма по статусам

status_counts = df_orders["status"].value_counts()

status_counts.plot(kind="bar")
plt.title("Количество заказов по статусам")
plt.xlabel("Статус")
plt.ylabel("Количество заказов")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 2. Диаграмма долей по статусам

status_shares = df_orders["status"].value_counts(normalize=True) * 100

status_shares.plot(kind="bar")
plt.title("Доли заказов по статусам")
plt.xlabel("Статус")
plt.ylabel("Доля (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# 3. Сравнение абсолютных значений и долей

status_comparison = pd.DataFrame({
    "Количество заказов": status_counts,
    "Доля (%)": status_shares
})

print("Сравнение количества заказов и долей:")
print(status_comparison)

# 4. Статус с наибольшим количеством заказов

print("Статус с наибольшим количеством заказов:")
print(status_counts.idxmax())

# 5. Типичная ошибка анализа

print("Типичная ошибка анализа:")
print(
    "Нельзя сравнивать только абсолютные значения, "
    "не учитывая доли заказов."
)
```
