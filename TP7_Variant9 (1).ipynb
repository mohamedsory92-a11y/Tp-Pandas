```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

orders_files = [
    "data/orders_2024_01.csv",
    "data/orders_2024_02.csv",
    "data/orders_2024_03.csv"
]

df_orders = pd.concat(
    [pd.read_csv(f) for f in orders_files],
    ignore_index=True
)

df_orders["order_datetime"] = pd.to_datetime(df_orders["order_datetime"])
df_orders["revenue"] = df_orders["price"] * df_orders["quantity"]

sns.pairplot(df_orders[["price", "quantity", "revenue"]])
plt.show()

corr = df_orders[["price", "quantity", "revenue"]].corr()
print(corr)

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Матрица корреляций")
plt.show()

corr_pairs = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
).stack()

print("Самая сильная положительная связь:")
print(corr_pairs.idxmax(), corr_pairs.max())

print("Самая слабая связь:")
print(corr_pairs.abs().idxmin())
```
